# 시리즈 미니퀘스트


*   .index: 인덱스 확인 → 기본값은 0부터 시작하는 RangeIndex
* pd.Series(): 리스트로 Series 생성
*   series['b']: 인덱스 'b'에 해당하는 값 출력

* .fillna(0): 결측값(NaN)을 0으로 대체



````python
# 01. Pandas의 Series를 리스트 [5, 10, 15, 20]을 사용하여 생성하고,
# Series의 인덱스를 확인하는 코드를 작성하세요.

import pandas as pd

series = pd.Series([5, 10, 15, 20]) # pd.Series() → 리스트로 Series 생성
print(series)
print(series.index)
````

````python
# 02. 다음 딕셔너리를 이용하여 Pandas Series를 생성하고,
# 인덱스를 활용하여 'b'의 값을 출력하는 코드를 작성하세요.

import pandas as pd

data = {'a': 100, 'b': 200, 'c': 300}
series = pd.Series(data)

print(series['b'])  # 인덱스 'b'에 해당하는 값 200 출력
````

````python
# 03. 다음 Pandas Series에 대해 결측값(NaN)을 확인하고,
# 모든 결측값을 0으로 채운 후 Series의 값을 출력하는 코드를 작성하세요.

import pandas as pd
series = pd.Series([1, 2, None, 4, None, 6])

print(series.isnull())  # 결측값 확인
filled = series.fillna(0) # 결측값을 0으로 채우기
print(filled)
````

# 필터링 미니퀘스트
* |: OR 연산자 (둘 중 하나만 True여도 선택)
* 조건문 작성할 때 조건마다 () 로 감싸주기!
* query(): 문자열로 조건을 작성하는 필터링 방법



> ❔ 언제 불리언을 쓰고 언제 쿼리를 쓰는게 좋은걸까?



````python
# 01. 주어진 데이터프레임에서 나이가 30 이상인 행만 필터링하는 코드를 작성하세요.

import pandas as pd

data = {'이름': ['홍길동', '김철수', '박영희', '이순신', '강감찬'],
        '나이': [25, 30, 35, 40, 45],
        '도시': ['서울', '부산', '서울', '대구', '부산']}

df = pd.DataFrame(data)

filtered_df = df[df['나이'] >= 30]

print(filtered_df)
````

````python
# 02. 주어진 데이터프레임에서 도시가 '서울'이거나 점수가 80점 이상인 데이터를 필터링하는 코드를 작성하세요.

import pandas as pd

data = {'이름': ['홍길동', '김철수', '박영희', '이순신', '강감찬'],
        '나이': [25, 30, 35, 40, 45],
        '도시': ['서울', '부산', '서울', '대구', '부산'],
        '점수': [85, 90, 75, 95, 80]}

df = pd.DataFrame(data)

filtered_df = df[(df['도시'] == '서울') | (df['점수'] >= 80)] # |: OR 연산자 (둘 중 하나만 True여도 선택), 조건마다 괄호() 필수!!!
print(filtered_df)
````

````python
# 03. 주어진 데이터프레임에서 query()를 사용하여 나이가 35 이상이고 점수가 80점 초과인 데이터를 필터링하는 코드를 작성하세요.

import pandas as pd

data = {'이름': ['홍길동', '김철수', '박영희', '이순신', '강감찬'],
        '나이': [25, 30, 35, 40, 45],
        '도시': ['서울', '부산', '서울', '대구', '부산'],
        '점수': [85, 90, 75, 95, 80]}

df = pd.DataFrame(data)

filtered_df = df.query('나이 >= 35 and 점수 > 80')
print(filtered_df)
````

# 그룹화 미니퀘스트

````python
# 01. 주어진 데이터프레임에서 groupby()를 사용하여 '부서'별로 급여의 합계를 구하는 코드를 작성하세요.

import pandas as pd

data = {
    '이름': ['홍길동', '김철수', '박영희', '이순신'],
    '부서': ['영업', '영업', '인사', '인사'],
    '급여': [5000, 5500, 4800, 5100]
}

df = pd.DataFrame(data)

grouped_df = df.groupby('부서')['급여'].sum()
print(grouped_df)
````

* groupby('부서') → 부서별로 데이터를 그룹화
* ['급여'] → 급여 열만 선택
* .sum() → 그룹별 합계 계산

````python
# 02. 아래 데이터프레임에서 groupby()와 agg()를 사용하여 '부서'별 급여의 합계(sum)와 평균(mean)을 계산하는 코드를 작성하세요.

import pandas as pd

data = {
    '이름': ['홍길동', '김철수', '박영희', '이순신', '강감찬', '신사임당'],
    '부서': ['영업', '영업', '인사', '인사', 'IT', 'IT'],
    '급여': [5000, 5500, 4800, 5100, 6000, 6200]
}

df = pd.DataFrame(data)

grouped_df = df.groupby('부서')['급여'].agg(['sum', 'mean'])  # agg(): 여러 집계 함수를 한 번에 적용
print(grouped_df)
````

````python
# 03. 부서별로 데이터를 그룹화한 뒤, 각 부서의 평균 급여가 5000 이상인 경우만 필터링하는 코드를 작성하시오.

import pandas as pd

data = {
    '이름': ['홍길동', '김철수', '박영희', '이순신', '강감찬', '신사임당'],
    '부서': ['영업', '영업', '인사', '인사', 'IT', 'IT'],
    '급여': [5000, 5500, 4800, 5100, 6000, 6200]
}

df = pd.DataFrame(data)

grouped_df = df.groupby('부서').filter(lambda x: x['급여'].mean() >= 5000)
print(grouped_df)
````

* **filter(lambda x: ...)**: 그룹별로 조건을 적용해 조건을 만족하는 그룹 전체를 반환
* **lambda**: 이름 없는 함수(익명 함수) 를 만드는 키워드
  * 일반 함수 vs lambda

    ```
      def 제곱(x):
        return x ** 2
    ```
    ```
      lambda x: x ** 2
    ```
    이 두개가 동일함!

  * 언제 사용?
    - 함수가 간단한 한 줄일 때
    - 딱 한 번만 쓸 때
    - `map()`, `filter()`, `apply()` 같은 함수랑 자주 같이 사용

# 병합 미니퀘스트

````python
# 01. 두 개의 데이터프레임을 고객ID를 기준으로 내부 조인(inner join)하여 공통된 데이터를 병합하는 코드를 작성하세요.

import pandas as pd

df1 = pd.DataFrame({'고객ID': [1, 2, 3], '이름': ['홍길동', '김철수', '이영희']})
df2 = pd.DataFrame({'고객ID': [2, 3, 4], '구매액': [10000, 20000, 30000]})

result = pd.merge(df1, df2, on='고객ID', how='inner') # on='고객ID': 기준 열 지정, how='inner': 양쪽 모두에 있는 데이터만 병합
print(result)
````

````python
# 02. 왼쪽 데이터프레임을 기준으로 병합(left join)을 수행하고, 구매액이 없는 경우 NaN을 유지하는 코드를 작성하세요.

import pandas as pd

df1 = pd.DataFrame({'고객ID': [1, 2, 3], '이름': ['홍길동', '김철수', '이영희']})
df2 = pd.DataFrame({'고객ID': [2, 3, 4], '구매액': [15000, 25000, 35000]})

result = pd.merge(df1, df2, on = '고객ID', how = 'left')  # NaN은 Pandas에서 자동으로 채워줌
print(result)
````

````python
# 03. 두 개 이상의 열을 기준으로 병합하고, 중복 열의 이름을 구분하도록 접미사(suffixes)를 지정하는 코드를 작성하세요.

import pandas as pd

df1 = pd.DataFrame({
    '고객ID': [1, 2, 3],
    '도시': ['서울', '부산', '대전'],
    '구매액': [10000, 20000, 30000]
})

df2 = pd.DataFrame({
    '고객ID': [1, 2, 3],
    '도시': ['서울', '부산', '광주'],
    '구매액': [15000, 25000, 35000]
})

result = pd.merge(df1, df2, on=['고객ID', '도시'], how='inner', suffixes=('_df1', '_df2'))
print(result)
````

* on=['고객ID', '도시']: 두 열을 동시에 기준으로 병합
* suffixes=('_df1', '_df2'): 중복 열(구매액)에 접미사 자동 부여

  * 구매액_df1: df1의 구매액
  * 구매액_df2: df2의 구매액


* 고객ID 3은 도시가 대전 vs 광주 로 달라서 inner join에서 제외

# 결측치 처리 미니퀘스트
* **isnull()**: 결측치면 True, 아니면 False
* isnull()**.sum()**: True를 1로 계산해서 열별 결측치 개수 합산
* **dropna()**: 결측치가 하나라도 포함된 행 전체 삭제

````python
# 01. 주어진 데이터프레임에서 결측치를 탐지하고, 열별 결측치 개수를 출력하는 코드를 작성하세요.

import pandas as pd
import numpy as np

data = {'이름': ['홍길동', '김철수', np.nan, '이영희'],
        '나이': [25, np.nan, 30, 28],
        '성별': ['남', '남', '여', np.nan]}

df = pd.DataFrame(data)

print(df.isnull())  # 결측치 확인
print(df.isnull().sum())  # 열별 결측치 개수
````

````python
# 02. 주어진 데이터프레임에서 결측치가 포함된 행을 삭제하고, 새로운 데이터프레임을 출력하는 코드를 작성하세요.

import pandas as pd
import numpy as np

data = {'이름': ['홍길동', '김철수', np.nan, '이영희'],
        '나이': [25, np.nan, 30, 28],
        '성별': ['남', '남', '여', np.nan]}

df = pd.DataFrame(data)

df_dropped = df.dropna()  # 결측치가 하나라도 포함된 행 전체 삭제
print(df_dropped)
````

````python
# 03. 결측치가 포함된 '나이' 열을 평균값으로 대체하고, 새로운 데이터프레임을 출력하는 코드를 작성하세요.

import pandas as pd
import numpy as np

data = {'이름': ['홍길동', '김철수', np.nan, '이영희'],
        '나이': [25, np.nan, 30, 28],
        '성별': ['남', '남', '여', np.nan]}

df = pd.DataFrame(data)

mean_age = df['나이'].mean()  # NaN 제외하고 평균 계산
df['나이'] = df['나이'].fillna(mean_age)  # 나이 열의 NaN만 평균값으로 대체
print(df)
````

# 피벗 미니퀘스트

````python
# 01. 주어진 데이터프레임을 pivot() 함수를 사용하여 날짜(날짜)를 행 인덱스로, 제품(제품)을 열로, 판매량(판매량)을 값으로 설정하는 코드를 작성하세요.

import pandas as pd

data = {
    '날짜': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02'],
    '제품': ['A', 'B', 'A', 'B'],
    '판매량': [100, 200, 150, 250]
}

df = pd.DataFrame(data)

pivot_df = df.pivot(index='날짜', columns='제품', values='판매량')
print(pivot_df)
````

````python
# 02. pivot_table()을 사용하여 주어진 데이터프레임에서 카테고리(카테고리)를 행 인덱스로, 제품(제품)을 열로 설정하고, 판매량(판매량)의 합계를 출력하는 코드를 작성하세요.

import pandas as pd

data = {
    '카테고리': ['전자', '가전', '전자', '가전'],
    '제품': ['A', 'B', 'A', 'B'],
    '판매량': [100, 200, 150, 250]
}

df = pd.DataFrame(data)

pivot_df = df.pivot_table(index = '카테고리', columns = '제품', values='판매량', aggfunc='sum') # aggfunc='sum': 같은 그룹의 값을 합계로 집계
print(pivot_df)
````

### pivot() vs pivot_table() 차이

| 구분 | pivot() | pivot_table() |
|---|---|---|
| 중복값 | ❌ 처리 불가 | ✅ 집계 가능 |
| 집계함수 | 없음 | aggfunc 지정 |
| 결측치 | 오류 발생 | NaN으로 처리 |

- **pivot()** → 중복 없는 깔끔한 데이터일 때
- **pivot_table()** → 중복값이 있거나 집계가 필요할 때

````python
# 03. 주어진 데이터프레임에서 여러 값을 동시에 피벗하여, pivot()을 사용해 날짜(날짜)를 행으로, 제품(제품)을 열로 설정하고, 판매량(판매량)과 이익(이익)을 동시에 피벗하는 코드를 작성하세요.

import pandas as pd

data = {
    '날짜': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02'],
    '제품': ['A', 'B', 'A', 'B'],
    '판매량': [100, 200, 150, 250],
    '이익': [20, 50, 30, 60]
}

df = pd.DataFrame(data)

df_pivot = df.pivot(index = '날짜', columns = '제품', values = ['판매량', '이익'])
print(df_pivot)
````

# 중복 제거 미니퀘스트
* drop_duplicates(): 모든 열이 동일한 행 제거
* duplicated():중복 행이면 True, 아니면 False

````python
# 01. 주어진 데이터프레임에서 기본적으로 중복된 행을 제거하는 코드를 작성하세요.

import pandas as pd

data = {
    '이름': ['김철수', '이영희', '김철수', '박민수'],
    '나이': [25, 30, 25, 40],
    '성별': ['남', '여', '남', '남']
}

df = pd.DataFrame(data)

new_df = df.drop_duplicates() # drop_duplicates(): 모든 열이 동일한 행 제거
print(new_df)
````

````python
# 02. 다음 데이터프레임에서 특정 열을 기준으로 중복을 제거한 후, 중복 여부를 확인하는 코드를 작성하세요.

import pandas as pd

data = {
    '제품': ['노트북', '태블릿', '노트북', '스마트폰'],
    '가격': [1500000, 800000, 1500000, 1000000],
    '카테고리': ['전자기기', '전자기기', '전자기기', '전자기기']
}

df = pd.DataFrame(data)

new_df = df.drop_duplicates(subset = ['제품', '가격'])  # 특정 열 기준 중복 제거
print(new_df)
print(new_df.duplicated())  # 중복 여부 확인, duplicated():중복 행이면 True, 아니면 False
````

````python
# 03. 다음 데이터프레임에서 중복된 모든 행을 삭제한 후, 남은 데이터를 저장하고 다시 불러오는 코드를 작성하세요.

import pandas as pd

data = {
    '학생': ['김민수', '박지현', '김민수', '이정훈'],
    '성적': [90, 85, 90, 88],
    '학교': ['A고', 'B고', 'A고', 'C고']
}

df = pd.DataFrame(data)

new_df = df.drop_duplicates() # 중복 행 삭제
new_df.to_csv('students.csv', index=False)  # to_csv('students.csv', index=False): CSV 파일로 저장, index=False: 인덱스 번호는 저장 안함
loaded_df = pd.read_csv('students.csv') # read_csv('students.csv'): 저장한 파일 다시 불러오기
print(loaded_df)
````

# 문자열 처리 미니퀘스트
* .str.lower(): 모든 문자열을 소문자로 변환
* str.strip(): 문자열에서 앞 뒤 공백 제거
* str.contains('Doe'): Doe 포함 여부 확인

````python
# 01. Pandas의 .str 접근자를 사용하여 아래 시리즈의 모든 문자열을 소문자로 변환하는 코드를 작성하세요.

import pandas as pd
data = pd.Series(["HELLO", "WORLD", "PYTHON", "PANDAS"])

result = data.str.lower()
print(result)
````

````python
# 02. 다음 데이터프레임에서 '이름' 컬럼의 문자열 앞뒤 공백을 제거하고,
# 특정 문자열 "Doe"가 포함된 행을 필터링하는 코드를 작성하세요.

import pandas as pd
df = pd.DataFrame({"이름": [" John Doe ", "Alice ", " Bob", "Charlie Doe "]})

df['이름'] = df['이름'].str.strip() # 앞뒤 공백 제거
filtered_df = df[df['이름'].str.contains('Doe')]  # "Doe" 포함 행 필터링
print(filtered_df)
````

````python
# 03. 아래 데이터프레임에서 '설명' 컬럼의 문자열을 공백을 기준으로 나누고, 각 단어의 첫 글자만 추출하여 새로운 컬럼 '약어'를 생성하는 코드를 작성하세요.

import pandas as pd
df = pd.DataFrame({"설명": ["빅데이터 분석", "데이터 과학", "머신 러닝", "딥 러닝"]})

df['약어'] = df['설명'].str.split().apply(lambda x: ''.join([word[0] for word in x]))  # 공백 기준으로 문자열 분리, 각 행에 함수 적용
print(df)
````

* str.split(): 공백 기준으로 문자열 분리
* apply(lambda x: ...): 각 행에 함수 적용
* ''.join(...): 추출한 첫 글자들을 하나의 문자열로 합치기
